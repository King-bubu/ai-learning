```python
from pathlib import Path
for path in Path("docs").glob("*.txt"):
    print(path)#「把符合 *.txt 这个形状的文件都给我」。（* 是通配符，代表任意字符。）
def load_chunks():
    """读 docs/ 下所有 txt,按空行切段,返回一个列表"""
    chunks = []
    for path in Path("docs").glob("*.txt"):
        text = path.read_text(encoding="utf-8")
        paras = text.split("\n\n")
        for i in paras:
            if i.strip() != "":
                chunks.append(i)
                print(i)
    return chunks
chunks=load_chunks()
print("共", len(chunks), "段")
print("第一段:", chunks[0])
print("最后一段:", chunks[-1])
        
```
